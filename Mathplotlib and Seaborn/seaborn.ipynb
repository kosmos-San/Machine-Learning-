```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
#Quantative data is visualize using distribution plots example:- Age,salary..etc
df=sns.load_dataset("tips")
df.head()
```

```python
df["size"].unique()
```

```python
#displot
plt.subplot(2,2,2)
sns.histplot(df["total_bill"],bins=20,kde=True)
plt.subplot(2,2,1)
sns.histplot(df["tip"],kde=True)
```

```python
#joint plot
sns.jointplot(x="total_bill",y="tip",data=df,kind="reg")
```

```python
#pairplot
sns.pairplot(df,hue="sex")
```

```python
#catagorical plot or qualatitive plot

sns.countplot(x=df["sex"],hue=df["smoker"])
```

```python
sns.barplot(x=df["sex"],y=df["total_bill"],estimator=np.sum)
```

```python
sns.boxplot(x=df["tip"],y=df["day"])
```

```python
sns.violinplot(x="day",y="total_bill",data=df)
```

```python
sns.stripplot(x="day",y="total_bill",data=df)
```

```python
sns.swarmplot(x="day",y="total_bill",data=df)
```

```python
sns.violinplot(x="tip",y="day",data=df)
sns.swarmplot(x="tip",y="day",data=df)
```

```python
flights=sns.load_dataset("flights")
flights.head()
```

```python
tips=sns.load_dataset("tips")
tips.head()
```

What is correlation
how data can be use to pridict another data.....so we find relationship between data.

```python
tipscorr=tips[['total_bill',"tip","size"]]
```

```python
tipscorr.corr()
```

```python
sns.heatmap(tipscorr.corr(),annot=True)
```

```python
sns.clustermap(tipscorr.corr())
```

```python
pvtflight=flights.pivot_table(values="passengers",index="month",columns="year")
```

```python
pvtflight
```

```python
sns.heatmap(pvtflight)
```
