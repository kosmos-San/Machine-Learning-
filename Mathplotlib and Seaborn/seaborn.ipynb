```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
#Quantative data is visualize using distribution plots example:- Age,salary..etc
df=sns.load_dataset("tips")
df.head()
```

```python
df["size"].unique()
```

```python
#displot
plt.subplot(2,2,2)
sns.histplot(df["total_bill"],bins=20,kde=True)
plt.subplot(2,2,1)
sns.histplot(df["tip"],kde=True)
```

```python
#joint plot
sns.jointplot(x="total_bill",y="tip",data=df,kind="reg")
```

```python
#pairplot
sns.pairplot(df,hue="sex")
```
