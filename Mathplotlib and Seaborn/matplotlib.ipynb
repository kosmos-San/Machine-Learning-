```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x=np.linspace(0,5,11)
x
```

```python
y=x**2
y
```

```python
#Basic command
plt.plot(x,y)
plt.title("hello")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

```python
z=np.arange(1,10,1)
u=z*9
```

```python
#subplot in matplotlib
plt.subplot(2,2,1)
plt.plot(x,y)
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.subplot(2,2,2)
plt.plot(z,u)


plt.subplot(2,2,3)
plt.plot(y,x)


plt.subplot(2,2,4)
plt.plot(u**4,z)

```

```python
# Object oriented way
fig=plt.figure(figsize=(10,10),dpi=100)
axis1=fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)

axis2=fig.add_axes([0.2,0.5,0.4,0.4])
axis2.plot(x,y)
```

```python
# Types of plot

#1 Scatter Plot
plt.scatter(x,y)
```

```python
#2 histogram
plt.hist(x)
```

```python
#3 box plot
plt.boxplot(x)
```

```python
fig=plt.figure()
ax=fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)

#save figure
plt.savefig("Basicplot.png")
```
