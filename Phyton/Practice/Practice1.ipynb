```python
# Q13. String Analysis

# Take a string from the user and count:

# uppercase letters
# lowercase letters
# digits
# spaces

# Example:

# Input: Hello World 123

str=input("Enter a string: ")
uucase=0
locase=0
dig=0
space=0
for i in str:
    if i.islower():
        locase+=1
    elif i.isupper():
        uucase+=1
    elif i.isdigit():
        dig+=1
    else:
        space+=1

print(str,uucase,locase,dig,space)
```

```python
# List Operations

# Given:

# numbers = [12, 5, 8, 21, 3, 15, 10]

# Find:

# Largest number
# Smallest number
# Sum of all numbers
# Number of even numbers
# Number of odd numbers

# Don't use max(), min(), or sum().

numbers=[12,5,8,21,3,15,10]
largest_number=numbers[0]
smallest_number=numbers[0]
sum=0
count_even=0
count_odd=0

for i in numbers:
    if i>largest_number:
        largest_number=i
    if i<smallest_number:
        smallest_number=i
    sum+=i
    if i%2==0:
        count_even=count_even+1
    else:
        count_odd=count_odd+1

print(largest_number,"is the largest number in list")
print(smallest_number,"is the smallest number in list")
print("sum of element in list is ",sum)
print("Even number in list are ",count_even)
print("Total odd number in list are ",count_odd)
```

```python
# Q11. Print 1 to N

# Write a recursive function that prints:

# 1 2 3 4 5 ... N

# Example:

# Input: 5
# Output: 1 2 3 4 5

def num(n):
    if n==0:
        return
    else:
        num(n-1)
        print(n)


print(num(100))
```

```python
# Q13. Factorial ⭐

# Calculate factorial using recursion.

# 5! = 120

def fact(num):
    if num==0:
        return 1
    else:
        return num*fact(num-1)


print(fact(5))
```

```python
# Q14. Sum of N Numbers

# Using recursion:

# Input: 5
# Output: 15

# Because:

# 5 + 4 + 3 + 2 + 1 = 15

def sum(num):
    if num==0:
        return 0
    else:
        return num+sum(num-1)

print(sum(25))
```

```python
# Q15. Power

# Write a recursive function:

# power(2, 5)

# Output:

# 32


def power(num,x):
    if x==1:
        return num
    else:
        return num*power(num,x-1)

print(power(2,4))
```

```python
# Q16. Count Digits

# Write a recursive function to count the digits of a number.

# Input: 58392
# Output: 5

def count(num):
    if num/10==0:
        return 0
    else:
        return 1+count(num//10)


print(count(1234))
```

```python
# Q17. Sum of Digits ⭐

# Write a recursive function:

Input=12345
# Output: 15

# Because:

# 1 + 2 + 3 + 4 + 5 = 15

def count_sum(x):
    if x//10<=0:
        return x
    else:
        digit=x%10
        return digit+count_sum(x//10)


print(count_sum(Input))
```

```python
# 2. Second Largest

# Find the second-largest distinct element without using sort().

# Example:

list=[10, 5, 8, 10, 3, 7]
max=list[0]
secondmax=float('-inf')

for i in list:
    if i>max:
        max=i

for i in list:
    if i>secondmax and i!=max:
        secondmax=i

print(secondmax)
print(max)

```

```python
# 3. Reverse an Array

# Reverse a list in-place.

# Example:

# [1, 2, 3, 4, 5]
# →
# [5, 4, 3, 2, 1]

# Try doing it without creating another list.

arr=[1,2,3,4,5]
i=0
j=len(arr)-1

while(i<j):
    temp=arr[i]
    arr[i]=arr[j]
    arr[j]=temp
    i=i+1
    j=j-1

print(arr)
```

```python
# 4. Rotate Array

# Rotate an array to the right by k positions.

# [1, 2, 3, 4, 5], k = 2
# → [4, 5, 1, 2, 3]

arr=[1,2,3,4,5] 
k=2
i=0
while(i<k):
    temp=arr[len(arr)-1]
    for z in range(len(arr)-1,-1,-1):
        arr[z]=arr[z-1]

    arr[0]=temp
    i+=1

print(arr)
```
