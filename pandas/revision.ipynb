```python
import pandas as pd

sales = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "customer": [
        "Aman", "Riya", "Kabir", "Neha", "Aman",
        "Sara", "Riya", "Arjun", "Neha", "Kabir"
    ],
    "city": [
        "Delhi", "Mumbai", "Delhi", "Pune", None,
        "Mumbai", "Delhi", "Pune", "Delhi", "Mumbai"
    ],
    "category": [
        "Electronics", "Clothing", "Electronics", "Furniture",
        "Clothing", "Electronics", "Furniture", "Clothing",
        "Electronics", "Furniture"
    ],
    "amount": [
        25000, 1500, 18000, None, 2500,
        32000, 7000, 3000, 22000, 9000
    ],
    "quantity": [2, 3, 1, 2, 5, 1, 2, 4, 2, 1],
    "rating": [5, 4, 5, 3, None, 5, 4, 3, 4, 5]
})
```

```python
#Q26. Replace missing values in `amount` with 0.
sales["amount"]=sales["amount"].fillna(0)
sales["amount"]
```

```python
#Q27. Fill missing `city` values with 'Unknown' and update the DataFrame.
sales["city"]=sales["city"].fillna("unkown")
```

```python
#Q28. Create a `total_items` column equal to `quantity` multiplied by 2.
sales["total_item"]=sales["quantity"]*2
sales.head()
```

```python
# Q29. Create a `discounted_amount` column equal to 90% of `amount`.
sales["discounted_amount"]=sales["amount"]*0.90
sales.head()
```

```python
#Q30. Drop the `rating` column into a new DataFrame without changing `sales`
sales["rating"]=sales["rating"].fillna(sales["rating"].mean())
rating2=sales.drop("rating", axis=1, inplace=False)
```

```python
#Q31. Calculate the total sales amount, ignoring missing values.
sales_amount=sales["amount"].sum()
sales_amount
```

```python
#Q32. Count the number of orders in each category.
sales.groupby("category")["order_id"].count()
```

```python
#Q33. Find the average amount for each category.
sales.groupby("category")["amount"].mean()
```

```python
#Find the maximum quantity for each city.
sales.groupby("city")["quantity"].max()
```

```python
#Q35. Sort orders by category alphabetically, then amount from highest to lowest.
sales.sort_values(by=["category","amount"],ascending=False)
```

```python
#Q36. Find the second-highest distinct order amount.
sales.sort_values(by="amount",ascending=False).iloc[2]
```

```python
#Q37. Find orders whose amount is greater than the average amount.
sales[sales["amount"]>sales["amount"].mean()]
```

```python
#Q38. Identify duplicate rows based on the customer column.
sales["customer"].duplicated()
```

```python
#Q40. Create a `sales_rank` column, ranking orders by amount from highest to lowest
sales["sales_rank"]=sales.rank()
sales.head()
```

```python
# Select orders from Delhi or Mumbai where the amount is greater than 10,000.
sales[(sales["city"].isin(["Delhi","Mumbai"]))& (sales["amount"]>10000)]
```

```python
#Q42. Display the customer and amount for orders with a rating of 5, sorted by amount in descending order.
sales[sales["rating"]==5][["customer","amount"]].sort_values(by="amount",ascending=False)
```

```python
#Q43. Find orders where either the city or rating is missing.
sales[(sales["city"].isnull())|(sales["rating"].isnull())]
sales.isnull().sum()
```

```python
#Create a summary with total amount, average amount, and order count for each category.

sales.groupby("category").agg(
    total_amount=("amount","sum"),
    average_amount=("amount","mean")
)
```
