```python
#Cleaning data


#removing all null value
df=employee.dropna()
df.shape
df.columns

df2=employee.dropna(subset=['city'])
df2.shape
```

```python
#filling those values
employee["age"] = employee["age"].fillna(employee["age"].mean())
employee.isnull().sum()

```

```python
employee["salary"]=employee["salary"].str.replace(",","",regex=False)
```

```python
employee["salary"]=pd.to_numeric(employee["salary"],errors="coerce")
employee['salary']=employee["salary"].fillna(employee["salary"].median())

employee.info()
```

```python
#department
employee["department"].value_counts()
```

```python
employee["department"]=employee["department"].fillna(employee["department"].mode()[0])
employee["city"]=employee["city"].fillna(employee["city"].mode()[0])
employee.isnull().sum()
```

```python
#Handling duplicates
employee.duplicated(subset=["employee_id"]).sum()
employee[employee.duplicated(subset=["employee_id"])]

employee=employee.drop_duplicates(subset=["employee_id"])

employee.duplicated(subset=["employee_id"]).sum()
employee
```

```python
#renaming a columns
employee.rename(columns={"name":"employee_name"})
```

```python
employee["department"]=employee["department"].str.strip().str.lower()
employee["department"].unique()
```

```python
#modifying data and creating new coloumns

#lets add annual salary columns
employee["salary"]=pd.to_numeric(employee["salary"],errors="coerce")
employee["annual_salary"]=employee["salary"] * 12
employee.head()
```

```python
#apply function()
def categorize(num):
    if num>5:
        return "Experienced"
    else:
        return "Learner"
```

```python
employee["category"]=employee["experience"].apply(categorize)
employee.head()
```

```python
#sorting and ranking 
employee["salary"]=pd.to_numeric(employee["salary"],errors="coerce")
employee["salary"]=employee["salary"].fillna(employee["salary"].mean())
employee.isnull().sum()
employee.sort_values(by="salary")
```

```python
employee["department"]=employee["department"].fillna(employee["department"].mode()[0])
```

```python
employee.sort_values(by="employee_id",ascending=False).head()
employee["department"]=employee["department"].str.strip().str.title()
employee["department"].unique()

employee.sort_values(by=["department","salary"],ascending=[True,False])
```

```python
employee["age"]=employee["age"].fillna(employee["age"].mean())
employee["age"]=employee["age"].astype(int)
```
